## Layer 4 — The CopilotKit runtime + the route map

The `__root.tsx` mounts the CopilotKit provider with:
```tsx
<CopilotKit runtimeUrl="/api/copilotkit/chat/completions" agent="gemini_hackathon_agent">
  <Outlet />
</CopilotKit>
```

The 8 routes that consume AGUI events:
- `/` — per-subnation home
- `/subjects` + `/subjects/$slug` — per-subject marimo embed
- `/agents` — chat panel
- `/find-resources` — cross-national resource discovery
- `/compare` — DuckDB-WASM leaderboard
- `/safeguarding` — per-subnation safeguarding policy
- `/archipelago` — 8 subnations side-by-side
- `/equivalency` — cross-jurisdiction equivalency UI

## The `useFrontendTool` / `useRenderTool` patterns

The 5 ADK tools surface to the UI via CopilotKit's `useRenderTool`:
- `lookup_outcome` → renders a syllabus-outcome card
- `retrieve_resources` → renders a resource list
- `find_similar_resources` → renders a cross-jurisdiction comparison table
- `retrieve_safeguarding` → renders a safeguarding policy panel
- `mark_answer` → renders a marking card with descriptor

Plus the `mcp_servers: { stitch: { url: STITCH_MCP_URL, transport: 'sse' } }` runtime config for the Stitch MCP integration.
